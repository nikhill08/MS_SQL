```python
import pyodbc
```

### Connect to SQL Server

```python
def connect_to_db():
    server = 'localhost'  
    database = 'firstDB'  
    username = 'sa'       
    password = 'nikhil123' 
    
    # Connection string using SQL Server Authentication
    connection_string = f'DRIVER={{ODBC Driver 18 for SQL Server}};SERVER={server};DATABASE={database};UID={username};PWD={password};TrustServerCertificate=yes;TrustedConnection=yes'

    try:
        # Establish the connection
        conn = pyodbc.connect(connection_string)
        print("Connection successful!")
        return conn
    except pyodbc.Error as e:
        print("Error in connection:", e)
        return None
    
conn = connect_to_db()

```

### Create Operation (Insert Data)

```python
def insert_employee(conn, name, age, department):
    cursor = conn.cursor()
    query = "INSERT INTO Employees (Name, Age, Department) VALUES (?, ?, ?)"
    
    try:
        cursor.execute(query, (name, age, department))
        conn.commit()
        print(f"Employee {name} inserted successfully!")
    except pyodbc.Error as e:
        print("Error while inserting:", e)
        
 
insert_employee(conn, 'Nikhil', 22, 'Comp')
    
 
```

### Read Operation (Select Data)

```python
def read_employees(conn):
    cursor = conn.cursor()
    query = "SELECT * FROM Employees"
    
    try:
        cursor.execute(query)
        rows = cursor.fetchall()
        for row in rows:
            print(f"ID: {row.EmployeeID}, Name: {row.Name}, Age: {row.Age}, Department: {row.Department}")
    except pyodbc.Error as e:
        print("Error while fetching data:", e)

read_employees(conn)
```

### Update Operation (Update Data)

```python
def update_employee_department(conn, employee_id, new_department):
    cursor = conn.cursor()
    query = "UPDATE Employees SET Department = ? WHERE EmployeeID = ?"
    
    try:
        cursor.execute(query, (new_department, employee_id))
        conn.commit()
        print(f"Employee ID {employee_id} department updated to {new_department}!")
    except pyodbc.Error as e:
        print("Error while updating:", e)

update_employee_department(conn, 6, 'ETC')
```

### Delete Operation (Delete Data)

```python
def delete_employee(conn, employee_id):
    cursor = conn.cursor()
    query = "DELETE FROM Employees WHERE EmployeeID = ?"
    
    try:
        cursor.execute(query, employee_id)
        conn.commit()
        print(f"Employee ID {employee_id} deleted successfully!")
    except pyodbc.Error as e:
        print("Error while deleting:", e)

delete_employee(conn, 5)
```
